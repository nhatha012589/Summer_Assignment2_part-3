## Finding 1: The distribution of SalePrice

```python
# Histogram of SalePrice
plt.figure(figsize=(8, 5))
sns.histplot(df["SalePrice"], bins=30, kde=True, color="steelblue")
plt.title("Distribution of House Sale Prices")
plt.xlabel("Sale Price ($)")
plt.ylabel("Number of Houses")
plt.show()
```

## Finding 2: Correlations with SalePrice

```python
numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()
correlation["SalePrice"].sort_values(ascending=False).head(12)
```

```python
selected_columns = ["SalePrice", "OverallQual", "GrLivArea",
    "GarageCars", "GarageArea", "TotalBsmtSF", "YearBuilt"]
plt.figure(figsize=(9, 6))
sns.heatmap(df[selected_columns].corr(),annot=True, cmap="coolwarm",fmt=".2f")
plt.title("Correlation Matrix of Selected Variables")
plt.tight_layout()
plt.show()
```

## Finding 3: Relationship between Overall Quality and SalePrice

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x="OverallQual", y="SalePrice", data=df)
plt.title("Overall Quality vs. Sale Price")
plt.xlabel("Overall Quality (1–10)")
plt.ylabel("Sale Price ($)")
plt.show()
```

## Finding 4: Relationship between Living Area and SalePrice

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x="GrLivArea", y="SalePrice", data=df)
plt.title("Above-ground Living Area vs. Sale Price")
plt.xlabel("Above-ground Living Area (square feet)")
plt.ylabel("Sale Price ($)")
plt.show()
```

## Finding 5: Average sale price by neighborhood 

```python
neighborhood_price = (df.groupby("Neighborhood")["SalePrice"].mean().sort_values(ascending=False).head(10))
neighborhood_price.plot(kind="bar", figsize=(9, 5), color="teal")
plt.title("Top 10 Neighborhoods by Average Sale Price")
plt.xlabel("Neighborhood")
plt.ylabel("Average Sale Price ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
